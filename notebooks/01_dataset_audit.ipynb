## Gate A: ds002721 audio recoverability
Run `src/01_audit_data.py` first, then use this notebook to review the audit CSV interactively, spot-check a few recovered clips against the original ratings, and visualize recovered/not_found/ambiguous counts.

```python
import pandas as pd
import matplotlib.pyplot as plt

audit = pd.read_csv("../data_raw/ds002721_stimulus_audit.csv")
print(f"Total clips: {len(audit)}")
print(audit["extraction_status"].value_counts())

plt.figure(figsize=(6, 4))
audit["n_participants"].hist(bins=range(1, audit["n_participants"].max() + 2))
plt.xlabel("Participants who heard this clip")
plt.ylabel("Number of clips")
plt.title("Rater coverage per clip (ds002721)")
plt.tight_layout()
plt.show()
```
